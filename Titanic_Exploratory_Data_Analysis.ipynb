```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/titanic.csv')

df.head()
```

```python

# Dataset information
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python

# Check duplicate records
print("Duplicate Rows:", df.duplicated().sum())

# Check column names
print("\nColumn Names:")
print(df.columns.tolist())

# Check passenger gender distribution
print("\nGender Distribution:")
print(df['Sex'].value_counts())

# Check passenger survival distribution
print("\nSurvival Distribution:")
print(df['Survived'].value_counts())

# Check passenger class distribution
print("\nPassenger Class Distribution:")
print(df['Pclass'].value_counts())
```

```python

# Set visualization style
sns.set_style("whitegrid")

# 1. Survival Count
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Survived')
plt.title('Survival Count')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')
plt.show()

# 2. Gender Distribution
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Sex')
plt.title('Passenger Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.show()

# 3. Passenger Class Distribution
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Pclass')
plt.title('Passenger Class Distribution')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')
plt.show()
```

```python

# 4. Age Distribution
plt.figure(figsize=(7,4))
sns.histplot(data=df, x='Age', bins=30, kde=True)
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

# 5. Fare Distribution
plt.figure(figsize=(7,4))
sns.histplot(data=df, x='Fare', bins=30, kde=True)
plt.title('Fare Distribution of Passengers')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.show()

# 6. Age Boxplot
plt.figure(figsize=(7,4))
sns.boxplot(data=df, x='Age')
plt.title('Age Outlier Detection')
plt.xlabel('Age')
plt.show()
```

```python

# 7. Survival by Gender
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Sex', hue='Survived')
plt.title('Survival by Gender')
plt.xlabel('Gender')
plt.ylabel('Passenger Count')
plt.show()

# 8. Survival by Passenger Class
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Pclass', hue='Survived')
plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Passenger Count')
plt.show()

# 9. Age vs Fare
plt.figure(figsize=(7,4))
sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived')
plt.title('Age vs Fare Based on Survival')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

```python

# Correlation Matrix
correlation = df.select_dtypes(include='number').corr()

# Correlation Heatmap
plt.figure(figsize=(10,6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Titanic Dataset')
plt.show()
```

```python

# Pairplot of numerical features
sns.pairplot(
    df[['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']].dropna(),
    hue='Survived'
)

plt.show()
```
